 If the item is "trap":
 Lose 5 coins
 Remove that item from the list

 If the item is "potion":
 Add a new "gold" to the end of the list

 Extra Rule (Important): If coins becomes less than 0, set it back to 0

 Final Output:

 Print the current coins


```python
import random
coins = 0
items = ["gold", "trap", "potion", "gold"]
pick = random.randint(0,3)
if pick == 0 or pick == 3:
    coins = coins + 10
    items.remove("gold")
    items.remove("gold")
    print(items)
if items == 1:
    coins = coins - 5
    items.remove("trap")
if items == 2:
    items.append("gold")

if coins < 0:
    coins = 0


print(f'you have {coins} coins')

```
