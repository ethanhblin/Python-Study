# 01 - From Function to Scope

```python
score = 10
def add_bonus(points):
    score = points + 5
    print(f"Inside the Function: {score}")
    return score

add_bonus(score)
print(f"Outside the function: {score}")
```

# 02 - What is Scope
Top Level = Global Scope 

Local Scope: any varieble in the function not callable outside a function

ex: message & topic

any print inside & outside of function is called the "Build-in Scope"

```python
course = "Python" #Global Scope

def show_lesson(topic): #Global Scope
    message = f"{course}: {topic}" 
    print(message) #Print = build-in scope
# print(message) THIS IS LOCAL SCOPE
show_lesson("Scope")#Global Scope
```

 ((((local scope) enclosing scope) global scope) build-in scope)

# 03 - Local Scope
for ex: Args & Kwargs

DEFINITION:

A local varieble is created inside a funcion and onlyt exist while the function is running. Once the function returns, the varieble is destroyed.

Note:

Code inside the function only exist when it is called, once executed, code gets destroyed but the function itself stays, so its an empty function

## Each call is independent

```python
def counter():
    count = 0
    count += 1
    return count
print(counter())
print(counter())
```

## Block scope does NOT exist in Python
Unlike languages such as Java or C++, Python does not create new scope inside `if`,`for`, or `while` blocks:



```python
if True:
    x = 10
print(x)

for i in range(5):
    square = i ** 2

print(i)
print(square)
```

Only **functions** and **classes** create new scope
