# 01 - From Function to Scope

```python
score = 10
def add_bonus(points):
    score = points + 5
    print(f"Inside the Function: {score}")
    return score

add_bonus(score)
print(f"Outside the function: {score}")
```

# 02 - What is Scope
Top Level = Global Scope 

Local Scope: any varieble in the function not callable outside a function

any print inside & outside of function is called the "Build-in Scope"

```python
course = "Python" #Global Scope

def show_lesson(topic): #Global Scope
    message = f"{course}: {topic}" 
    print(message) #Print = build-in scope
# print(message) THIS IS LOCAL SCOPE
show_lesson("Scope")#Global Scope
```

## Each call is independent

```python
def counter():
    count = 0
    count += 1
    return count
print(counter())
print(counter())
```

## Block scope does NOT exist in Python
Unlike languages such as Java or C++, Python does not create new scope inside `if`,`for`, or `while` blocks:



```python
if True:
    x = 10
print(x)

for i in range(5):
    square = i ** 2

print(i)
print(square)
```

Only **functions** and **classes** create new scope

# 05 -  Enclosing Scope

- When you define a function insdide another function, the inner function can access varieble from the **encloing**(outer) function.

```python
def outer():
    message = "Hello from outer" # enclosing scope varieble

    def inner():
        print(message)           # Accesses enclosing varieble

    inner()

outer()   # output: Hello from outer
```

```python
def level_one():
    x = "one"
    def level_two():
        y = "two"
        def level_three():
            print(x,y)
        level_three()
    level_two()
level_one()
```

# 06 - Built-in Scope
Pre-programed 

```python
def show_length(text):
    return len(text)

print(show_length("Python"))
```

 - `show_length` is **Global** because the function is defined at module level.
 - `text` is **Local** because it is a parameter of `show_length()`.
 - `len` and `print` are **Built-in** because Python provides them automatically
