🔹 Task

Write a function called:

class_scores
🔹 It must:
Take a class name as a normal argument
Take any number of student scores using *args
Take extra info using **kwargs (like teacher, room, subject)
🔹 What it should do:

When called, it should print:

1. Class name
2. All scores (clean format)
3. Highest score
4. Lowest score
5. Extra info (from kwargs)

class_scores(

🔥 Rules
Must use a function

Must use **kwargs
args will automatically become a tuple

```python
def class_scores(class_name, *args, **kwargs, ):
    print(class_name)
    print(kwargs)
    print(f'This is the class scores. {args}')
    average = sum(args) / len(args)
    print(f'The class average is {average}')
    largest = args[0]
    for num in args:
        if num > largest:
            largest = num
        
    print(f'The highest grade is {largest}')
    smallest = args[0]
    for num in args:
        if num < smallest:
            smallest = num
    print(f'The lowest grade is {smallest}')



class_scores("ethans class",35, 56, 75, 84, 98, 97, 91, 79, 88, teacher = "Lin")
```

# 13 - Nested Function

```python
def format_reciept(item,price, tax_rate):
    def format_money(amount):
        return f"${amount:.2f}"

    tax = price * tax_rate
    total = price + tax
    return (
        f"{item} | Price: {format_money(price)} |"
        f"Tax: {format_money(tax)} |"
        f"Total: {format_money(total)}"
    )
print(format_reciept("Notebook", 10, 0.05))
```

# A5 - Student grade analyzer

```python

def average_grade(grades):
    print(grades)
    amount_of_student_grades = 0
    total_amount_of_grades = 0
    for individual_grade in grades:
        total_amount_of_grades += individual_grade
        amount_of_student_grades += 1
    final_average_grade_of_students = total_amount_of_grades / amount_of_student_grades
    return final_average_grade_of_students
    

def letter_grade(my_average):
    if my_average >= 90:
        print("Letter Grade : A")
    elif my_average >= 80 and my_average < 90:
        print("Letter Grade : B")
    elif my_average >= 70 and my_average < 80:
        print("Letter Grade : C")
    elif my_average >= 60 and my_average < 70:
        print("Letter Grade : D")
    elif my_average < 60 :
        print("Letter Grade : F")
        
grades = [88,92,76,95]
my_average = average_grade(grades)
print(my_average)
letter_grade(my_average)

```

# A6 -  Merging Ranges

```python
def merge_range(range):
    range.sort()
    print(range)
    first_num = 10000000
    sec_num = 0
    for range1 in range:

        if first_num > range1[0]:
            first_num = range1[0]
        if sec_num < range1[1]:
            sec_num = range1[1]
    print(first_num, sec_num)



merge_range([(1, 3),(5,9),(2,7)])
```
