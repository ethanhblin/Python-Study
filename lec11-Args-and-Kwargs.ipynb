# 01 -  **`*args`** -- Varieble Positional Arguments

**Basic Usage**


```python
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3))
print(add_all(5, 10, 15, 20))
print(add_all())
```

**The Name "args" is a convention**

```python
def multiply(*numbers):
    result = 1
    for n in numbers:
        result *= n
    return result

multiply(2,3,4)
```

**args is a tuple**

```python
def show_type(*args):
    print(type(args))
    print(args)
    print(f"First:{args[0]}")
    print(f"Rest: {args[1: ]}")

show_type(1,2,3)
```

**Mixing Regular Parameter with** ***args**

```python
def tag(name, *content):
    inner = "7".join(content)
    return f"<{name}><{inner}></{name}>"
tag("p","Hello"," ","world")
```

**Practical Use: Flexible Math Functions**

```python
def average(*args):
    if not args:
        raise ValueError("Wrong")
    return sum(args) / len(args)

print(average(85,90,78))
print(average(100))
print(average(0))
```

## A1 - joining words

```python
def join_words(seperator, *words):
    return seperator.join(words)


print(join_words("-", "scope", "args", "kwargs"))

print(join_words(", "))
```
